# Data Preprocessing 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,f1_score,precision_score,recall_score,confusion_matrix
```

```python
data = pd.read_csv(r"C:\Users\ASUS\Desktop\Mini Project\Ensemble_prediction_system\deepfake_features.csv")
data.head()
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data.info()
```

```python
data.describe()
```

```python
X = data.drop('label',axis=1)
y = data['label']
```

```python
display(X,y)
```

```python
x_train,x_test,y_train,y_test = train_test_split(X,y,test_size=0.1,random_state=42)
```

```python
display(x_train,x_test,y_train,y_test)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

# Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model_rf = RandomForestClassifier(n_estimators=100, random_state=42)
model_rf.fit(x_train,y_train)
```

```python
y_pred = model_rf.predict(x_test)
```

```python
accuracy = accuracy_score(y_test,y_pred)
precision = precision_score(y_test,y_pred)
recall = recall_score(y_test,y_pred)
f1 = f1_score(y_test,y_pred)
```

```python
print("Accuracy:",accuracy)
print("Precision:",precision)
print("Recall:",recall)
print("F1-score:",f1)
```

```python
importance = model_rf.feature_importances_
feature_names = X.columns

feat_imp = pd.Series(importance, index=feature_names).sort_values(ascending=False)
print(feat_imp)
```

```python
print("Accuracy:",accuracy)
print("Precision:",precision)
print("Recall:",recall)
print("F1-score:",f1)
```

# SVM (RBF kernel)

```python
from sklearn.svm import SVC
```

```python
svm = SVC(
    kernel='rbf',   
    C=1.0,
    gamma='scale',
    probability=True
)
svm.fit(x_train_scaled, y_train)
```

# Comparative Analysis

```python
results = []

def evaluate_model(name, y_true, y_pred):
    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred),
        "Recall": recall_score(y_true, y_pred),
        "F1 Score": f1_score(y_true, y_pred)
    })


# Random Forest
evaluate_model("Random Forest", y_test, y_pred)

# XGBoost
evaluate_model("XGBoost", y_test, y_pred_xg)

# LightGBM
evaluate_model("LightGBM", y_test, y_pred_lgbm)

# Logistic Regression
evaluate_model("Logistic Regression", y_test, y_pred_lr)

# SVM
evaluate_model("SVM (RBF)", y_test, y_pred_svm)


results_df = pd.DataFrame(results)
results_df = results_df.sort_values(by="F1 Score", ascending=False)

results_df.style.highlight_max(color='lightgreen')
```

# Meta-Meta Ensemble

```python
from sklearn.ensemble import VotingClassifier

voting = VotingClassifier(
    estimators=[
        ('lr', lr),
        ('xgb', model_xg),
        ('svm', svm)
    ],
    voting='soft'   # IMPORTANT
)

voting.fit(x_train_scaled, y_train)
y_pred_voting = voting.predict(x_test_scaled)
```

```python
accuracy = accuracy_score(y_test,y_pred_voting)
precision = precision_score(y_test,y_pred_voting)
recall = recall_score(y_test,y_pred_voting)
f1 = f1_score(y_test,y_pred_voting)

print("Accuracy:",accuracy)
print("Precision:",precision)
print("Recall:",recall)
print("F1-score:",f1)
```
